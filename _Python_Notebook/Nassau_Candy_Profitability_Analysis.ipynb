```python
#GROSS MARGIN %

df["Gross Margin (%)"] = (
    df["Gross Profit"] / df["Sales"]
) * 100
```

```python
# PROFIT PER UNIT

df["Profit per Unit"] = (
    df["Gross Profit"] / df["Units"]
)
```

```python
# REVENUE CONTRIBUTION 

df["Revenue Contribution (%)"] = (
    df["Sales"] / df["Sales"].sum()
) * 100
```

```python
# PROFIT CONTRIBUTION

df["Profit Contribution (%)"] = (
    df["Gross Profit"] / df["Gross Profit"].sum()
) * 100
```

```python
df.head()
```

```python
product_profit = df.groupby(
    ["Product ID", "Product Name"]
).agg(
    Sales=("Sales", "sum"),
    Units=("Units", "sum"),
    Gross_Profit=("Gross Profit", "sum"),
    Cost=("Cost", "sum")
).reset_index()
```

```python
product_profit["Gross Margin (%)"] = (
    product_profit["Gross_Profit"] /
    product_profit["Sales"]
) * 100
```

```python
product_profit.sort_values(
    "Gross_Profit",
    ascending=False
).head(10)
```

```python
division_analysis = df.groupby("Division").agg(
    Sales=("Sales", "sum"),
    Gross_Profit=("Gross Profit", "sum"),
    Cost=("Cost", "sum"),
    Units=("Units", "sum")
).reset_index()
```

```python
division_analysis["Gross Margin (%)"] = (
    division_analysis["Gross_Profit"] /
    division_analysis["Sales"]
) * 100
```

```python
# Median values as thresholds
sales_threshold = product_profit["Sales"].median()
margin_threshold = product_profit["Gross Margin (%)"].median()

sales_threshold, margin_threshold
```

```python
def classify_product(row):
    if row["Sales"] >= sales_threshold and row["Gross Margin (%)"] >= margin_threshold:
        return "High Sales - High Margin"
    elif row["Sales"] >= sales_threshold and row["Gross Margin (%)"] < margin_threshold:
        return "High Sales - Low Margin"
    elif row["Sales"] < sales_threshold and row["Gross Margin (%)"] >= margin_threshold:
        return "Low Sales - High Margin"
    else:
        return "Low Sales - Low Margin"


product_profit["Profitability Segment"] = product_profit.apply(
    classify_product,
    axis=1
)

product_profit.head()
```

```python
product_profit["Profitability Segment"].value_counts()
```

```python
high_sales_low_margin = product_profit[
    product_profit["Profitability Segment"] == "High Sales - Low Margin"
].sort_values(
    "Sales",
    ascending=False
)

high_sales_low_margin
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=product_profit,
    x="Cost",
    y="Gross Margin (%)",
    hue="Profitability Segment"
)

plt.title("Cost vs Gross Margin")
plt.xlabel("Cost")
plt.ylabel("Gross Margin (%)")

plt.show()
```

```python
profit_pareto = product_profit.sort_values(
    "Gross_Profit",
    ascending=False
).copy()
```

```python
profit_pareto["Cumulative Profit"] = (
    profit_pareto["Gross_Profit"].cumsum()
)
```

```python
total_profit = profit_pareto["Gross_Profit"].sum()

profit_pareto["Cumulative Profit (%)"] = (
    profit_pareto["Cumulative Profit"] / total_profit
) * 100
```

```python
products_for_80_profit = (
    profit_pareto["Cumulative Profit (%)"] <= 80
).sum()

products_for_80_profit
```

```python
revenue_pareto = product_profit.sort_values(
    "Sales",
    ascending=False
).copy()

revenue_pareto["Cumulative Sales"] = (
    revenue_pareto["Sales"].cumsum()
)

total_sales = revenue_pareto["Sales"].sum()

revenue_pareto["Cumulative Sales (%)"] = (
    revenue_pareto["Cumulative Sales"] / total_sales
) * 100
```

```python
products_for_80_revenue = (
    revenue_pareto["Cumulative Sales (%)"] <= 80
).sum()

products_for_80_revenue
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    range(1, len(profit_pareto) + 1),
    profit_pareto["Cumulative Profit (%)"]
)

plt.axhline(80, linestyle="--")

plt.title("Pareto Analysis - Cumulative Profit Contribution")
plt.xlabel("Number of Products")
plt.ylabel("Cumulative Profit (%)")

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    range(1, len(revenue_pareto) + 1),
    revenue_pareto["Cumulative Sales (%)"]
)

plt.axhline(80, linestyle="--")

plt.title("Pareto Analysis - Cumulative Revenue Contribution")
plt.xlabel("Number of Products")
plt.ylabel("Cumulative Revenue (%)")

plt.show()
```

```python
monthly_analysis = df.groupby(
    df["Order Date"].dt.to_period("M")
).agg(
    Sales=("Sales", "sum"),
    Gross_Profit=("Gross Profit", "sum")
).reset_index()

monthly_analysis["Gross Margin (%)"] = (
    monthly_analysis["Gross_Profit"] /
    monthly_analysis["Sales"]
) * 100

monthly_analysis
```

```python
margin_volatility = monthly_analysis["Gross Margin (%)"].std()

margin_volatility
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    monthly_analysis["Order Date"].astype(str),
    monthly_analysis["Gross Margin (%)"],
    marker="o"
)

plt.title("Monthly Gross Margin Trend")
plt.xlabel("Month")
plt.ylabel("Gross Margin (%)")

plt.xticks(rotation=45)

plt.show()
```
